```python
from transformers import (
    AutoTokenizer
)
```

```python
xmistral_model_name = 'Hannibal046/xrag-7b'
gemma2_model_name = 'google/gemma-2-2b-it'
```

```python
xmistral_tokenizer = AutoTokenizer.from_pretrained(
    xmistral_model_name,
    padding_side = 'left',
    add_eos_token=False, ## import to include this!
    use_fast=False,
)
```

```python
gemma2_toknenizer = AutoTokenizer.from_pretrained(gemma2_model_name)
```

```python
vocab1 = set(xmistral_tokenizer.get_vocab().keys())
vocab2 = set(gemma2_toknenizer.get_vocab().keys())
```
