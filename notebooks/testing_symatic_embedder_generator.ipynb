```python
documents_list = [
    'In January 2009 President Barack Obama restored US funding to UNFPA, saying in a public statement that he would "look forward to working with Congress to restore US financial support for the UN Population Fund. By resuming funding to UNFPA, the US will be joining 180 other donor nations working collaboratively to reduce poverty, improve the health of women and children, prevent HIV/AIDS and provide family planning assistance to women in 154 countries',
    'The United States rejoined a collaborative effort with 180 other donor nations to address several global challenges. In January 2009, President Barack Obama announced the reinstatement of U.S. funding for the UN Population Fund (UNFPA), stating he would work with Congress to formalize the financial support. This renewed partnership focuses on reducing poverty, improving health for women and children, preventing HIV/AIDS, and providing family planning assistance to women across 154 countries.'
]
```

```python
embeds = prepare_retrieval_embeds(documents_list, model, tokenizer, batch_size=2)
```

```python
import torch
import torch.nn.functional as F
```

```python
stacked_embeddings = torch.stack([embed for embed in embeds])
```

```python
embeddings_mean = torch.mean(stacked_embeddings, dim=0)
```

```python
cosine_sim = F.cosine_similarity(stacked_embeddings[0], stacked_embeddings[1], dim=0)
```
